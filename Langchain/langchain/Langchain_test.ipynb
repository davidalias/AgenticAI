```python
from langchain.chat_models import ChatOpenAI
from langchain.schema import HumanMessage

llm = ChatOpenAI(model="gpt-4o-mini", temperature=0.7)  # Use ChatOpenAI for chat models
prompt = "Write a poem about python and ai"
# Convert the prompt string to a HumanMessage object
messages = [HumanMessage(content=prompt)]
print(llm(messages)) # Pass the messages list to llm
```
