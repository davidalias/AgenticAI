```python
import sys
import os
from autogen import ConversableAgent, GroupChat, GroupChatManager

#Add parent directory to sys.path
sys.path.append(os.path.abspath(".."))

from llm_config import llm_config
```

Define Cake baking agents

```python
ingredient_agent = ConversableAgent(
    "IngredientAgent",
    llm_config = llm_config,
    system_message = """You are responsible for gathering ingredients for any cake requested""" #share instructions
)
#description method helps the group chat agent to understand the skills of each agent
ingredient_agent.description = "Gather ingredients for the requested cake" #this description will be referred by groupchat manager to select the relevant agent during "auto" method
```

```python
mixing_agent = ConversableAgent(
    name= "MixingAgent",
    llm_config=llm_config,
    system_message= """You mix all the ingredients provided into a batter"""
)

mixing_agent.description = "Mix ingredients to prepare cake batter"
```

```python
baking_agent = ConversableAgent(
    name= "BakingAgent",
    llm_config=llm_config,
    system_message= """You bake the batter at the appropriate temperature"""
)

baking_agent.description = "Bake the cake batter in an oven"
```

```python
decorating_agent = ConversableAgent(
    name= "DecoratingAgent",
    llm_config=llm_config,
    system_message= """You decorate the cake based on the requested flavor"""
)

decorating_agent.description = "Decorate the cake with relevant toppings"
```

```python
tasting_agent = ConversableAgent(
    name= "TastingAgent",
    llm_config=llm_config,
    system_message= """You taste the final cake and give feedback"""
)

tasting_agent.description = "Taste the final cake and provide review"
```

```python
home_baker_agent = ConversableAgent(
    name="HomeBakerAgent",
    llm_config=llm_config,
    system_message="""You are a home baker trying to bake cakes with the help of other agents"""
)
```

```python
group_chat = GroupChat(
    agents= [
        ingredient_agent,
        mixing_agent,
        baking_agent,
        decorating_agent,
        tasting_agent
    ],
    messages=[],
    max_round= 6 #each agent once
)
```

```python
group_chat_manager = GroupChatManager(
    groupchat=group_chat,
    llm_config=llm_config
)
```

```python
chat_result = home_baker_agent.initiate_chat(
    group_chat_manager,
    message= "I want to bake a chocolate cake",
    summary_method="reflection_with_llm"
)
```

Agents sending instructions with each other

```python
group_chat_intros = GroupChat(
    agents= [
        ingredient_agent,
        mixing_agent,
        baking_agent,
        decorating_agent,
        tasting_agent
    ],
    messages=[],
    max_round= 6, #each agent once
    send_introductions= True #agents can send instructions between each other
)
```

```python
group_chat_manager_with_intros = GroupChatManager(
    groupchat=group_chat_intros,
    llm_config=llm_config
)
```

```python
chat_result = home_baker_agent.initiate_chats(
  [ 
      {
       "recipient" : group_chat_manager_with_intros,
       "message" : "I want to bake a chocolate cake"
      },
      {
          "recipient" : group_chat_manager_with_intros,
          "message" : "Now I want to bake a vanilla cake"
      }
   ]
)
```
