```python
import sys
import os
import tempfile
from autogen import ConversableAgent

#Add parent directory to sys.path
sys.path.append(os.path.abspath(".."))

from llm_config import llm_config
```

```python
number_agent = ConversableAgent(
    name= "Number_Agent",
    system_message= "You return each number that I share with you, one number each line",
    llm_config=llm_config,
    human_input_mode="NEVER"
)
```

```python
#The adder agent adds 1 to each number it receives
adder_agent = ConversableAgent(
    name="Adder_Agent",
    system_message="You add 1 to each number I give, and return me the new number, with one number on each line",
    llm_config=llm_config,
    human_input_mode="NEVER"
)
```

```python
#Mulitplier agent multiplies each number it receives by 2
multiplier_agent = ConversableAgent(
    name="Multiplier_Agent",
    system_message= " You multiply each number I provide with by 2, and return the new number, with one number per line",
    llm_config=llm_config,
    human_input_mode="NEVER"
)
```

```python
#Subtractor agent subtracts each number it receives by 1
subtractor_agent = ConversableAgent(
    name="Subtractor_Agent",
    system_message= " You subtract each number I provide with by 1, and return the new number, with one number per line",
    llm_config=llm_config,
    human_input_mode="NEVER"
)
```

```python
#Divider agent divides each number it receives by 2
divider_agent = ConversableAgent(
    name="Divider Agent",
    system_message= " You divide each number I provide with by 2, and return the new number, with one number per line",
    llm_config=llm_config,
    human_input_mode="NEVER"
)
```

```python
chat_result = number_agent.initiate_chats(
    [
        {
            "recipient" : adder_agent,
            "message": "14",
            "max_turns": 1,
            "summary_method": "last_msg"
        },
        {
            "recipient" : multiplier_agent,
            "message": "These are my numbers",
            "max_turns": 1,
            "summary_method": "last_msg"
        },
        {
            "recipient" : subtractor_agent,
            "message": "These are my numbers",
            "max_turns": 1,
            "summary_method": "last_msg"
        },
        {
            "recipient" : divider_agent,
            "message": "These are my numbers",
            "max_turns": 1,
            "summary_method": "last_msg"
        }
    ]
)
```
