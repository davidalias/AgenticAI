```python
import sys
import tempfile
import os
from autogen.agentchat import ConversableAgent
from autogen.coding import DockerCommandLineCodeExecutor
```

```python
#add parent directory to sys.path
sys.path.append(os.path.abspath(".."))
from llm_config import llm_config
```

```python
temp_dir = tempfile.TemporaryDirectory() # temporary directory to store the code files

#Docker commandline executor
executor = DockerCommandLineCodeExecutor(
    image = 'python:3.12-slim', # execute code using the given docker image name
    timeout = 10, #timeout for each code execution in seconds
    work_dir = temp_dir.name #use the temporary directory to store the code files
)
```

```python
code_executor_agent_using_docker = ConversableAgent(
    "code_executor_agent_docker",
    llm_config = False,
    code_execution_config = {"executor": executor}, #use docker command line code executor
    human_input_mode ="NEVER"
)
```

```python
code_writer_system_message = """You are a helpful AI assistant.
Solve tasks using your coding and language skills.
In the following cases, suggest python code (in a python coding block) for the user to execute:
1. When you need to collect info, use the code to output the info you need
2. When you need to perform some tasks with code, use the code to perform the task and output the result. Finish the task smartly.
When you find an answer, verify the answer carefully. Include verifiable evidence in your response if possible
Reply 'TERMINATE' in the end when everything is done"""
```

```python
code_writer_agent = ConversableAgent(
    "code_writer_agent",
    system_message = code_writer_system_message,
    llm_config = llm_config,
    code_execution_config = False
)
```

```python
chat_result = code_executor_agent_using_docker.initiate_chat(
    code_writer_agent,
    message = "Write python code to calculate the 14th Fibonacci number. Install required dependencies if required."
)
```
