```python
from autogen_agentchat.agents import AssistantAgent
from autogen_agentchat.ui import Console
from autogen_ext.models.openai import OpenAIChatCompletionClient
import os 
import sys
sys.path.append(os.path.abspath(".."))
from dotenv import load_dotenv
load_dotenv()
```

```python
# Global counter state
counter = 0
```

```python
def increment_counter() -> str:
    """Increment the counter value by 1 and return the current value"""
    global counter
    counter += 1
    return f"The current counter value is {counter}"
```

```python
def current_value()->str:
    """Return the current counter value"""
    global counter
    return f"The current value is {counter}"
```

```python
model_client = OpenAIChatCompletionClient(
    model='gpt-3.5-turbo',
)
```

```python
agent = AssistantAgent(
    name = 'agent',
    model_client=model_client,
    tools=[increment_counter, current_value],
    max_tool_iterations=5, # Allow up to 5 tool call iterations
    reflect_on_tool_use=True # Get a final summary after tool calls
)
```

```python
agent_run = agent.run_stream(task="Increment the counter 3 times and share the final value")
await Console(agent_run)
```
