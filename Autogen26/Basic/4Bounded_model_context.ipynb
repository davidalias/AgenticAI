```python
from autogen_agentchat.agents import AssistantAgent
from autogen_core.model_context import BufferedChatCompletionContext
from autogen_ext.models.openai import OpenAIChatCompletionClient
import os 
import sys
sys.path.append(os.path.abspath(".."))
from dotenv import load_dotenv
load_dotenv()
```

```python
model_client = OpenAIChatCompletionClient(
    model='gpt-3.5-turbo'
)
```

```python
# Create a model context that only keeps the last 2 messages (1 user + 1 assistant).
model_context = BufferedChatCompletionContext(buffer_size=2)
```

```python
# Create an AssistantAgent instance with the model client and context.
agent = AssistantAgent(
        name="assistant",
        model_client=model_client,
        model_context=model_context,
        system_message="You are a helpful assistant.",
    )
```

```python
result = await agent.run(task="Who is the captain of Indian cricket team?")
print(result.messages[-1].content)
```

```python
result = await agent.run(task = "Elaborate on it")
print(result.messages[-1].content)
```

```python
result = await agent.run(task = "When did he started playing for India?")
print(result.messages[-1].content)
```
