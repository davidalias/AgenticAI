```python
from autogen_agentchat.agents import AssistantAgent, UserProxyAgent
from autogen_agentchat.conditions import MaxMessageTermination, TextMentionTermination
from autogen_ext.models.openai import OpenAIChatCompletionClient
from autogen_agentchat.ui import Console
from autogen_agentchat.teams import RoundRobinGroupChat
import os
import sys
sys.path.append(os.path.abspath(".."))
from dotenv import load_dotenv
load_dotenv()
```

```python
model_client = OpenAIChatCompletionClient(
    model="gpt-3.5-turbo"
)
```

```python
writer = AssistantAgent(
    name = 'writer',
    model_client= model_client,
    system_message="You are a writer",
    model_client_stream=True
)
```

```python
reviewer = AssistantAgent(
    name = 'reviewer',
    system_message="Read the input article, provide feedback and suggest improvements if required",
    model_client= model_client,
    model_client_stream=True
)
```

```python
user = UserProxyAgent(
    name = 'user'
)
```

```python
 # Maximum 1 round of review and revision.
inner_termination = MaxMessageTermination(max_messages=4)
```

```python
# The outer-loop termination condition that will terminate the team when the user types "exit".
outer_termination = TextMentionTermination("exit", sources=["user"])
```

```python
team = RoundRobinGroupChat(
    [RoundRobinGroupChat([writer, reviewer], termination_condition=inner_termination), 
     user],
     termination_condition=outer_termination
)
```

```python
await Console(team.run_stream(task = "Write an article about the impact of war on Earth's life"))
```
